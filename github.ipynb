# Chuleta de Comandos de github

Flujo habitual para subir cambios

```python
git add . 

git commit -m "Mensaje_commit"

git push origin rama_destino
```

Crear Rama, a partir de la que estas

Recuerda que para que se quede guardada en remoto y no en local, hace falta hacer un push despues de crearla

```python
git branch -m rama_actual rama_nueva
```

Eliminar Rama

```python
git push origin --delete rama
```

Cambiar de Rama

```python
git checkout rama
```

Actualizar main a una rama

```python
1. Entrar en main
git checkout main o git switch main

2. Hacer merge de rama a main
git merge rama

3. Subir los cambios a GitHub
git push origin main

4. Volver a tu rama
git checkout rama o git switch rama
```

Ver las diferencias entre dos ramas

```python
git diff rama1..rama2
```
