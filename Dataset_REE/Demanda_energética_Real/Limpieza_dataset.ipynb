## Codigo para la lectura/limpieza de los datos de la demanda energética real

### 1 Leer y mostrar informacion acerca del DF

```python
import pandas as pd

nombre_csv = "demanda_real_completa2021.csv"

# Leer el archivo CSV
df = pd.read_csv(nombre_csv, sep=';')

# Mostrar información básica del DataFrame
print("Información del DataFrame:")
print(f"Dimensiones: {df.shape}")
#print(f"\nDataFrame procesado. Total de registros: {len(df)}") ya lo devuele .shape

# Mostrar estadísticas descriptivas de los valores (IMPORTANTE MUY ÚTIL)
print(f"\nEstadísticas de 'value':\n{df['value'].describe()}")

print(f"\nPrimeras filas:\n{df.head()}")

#print(f"\nTipos de datos:\n{df.dtypes}")
```

### 3 Eliminar las columnas sin datos o innecesarias

```python
#Como geoid y geoname no tienen valores, se pueden elimnar
df.drop('geoid', axis=1)
df.drop('geoname', axis=1)
df.drop('name', axis=1)

#Tambien se puede hacer asi:
del df['geoid']
del df['geoname']
del df['name']



print(f"\nPrimeras filas:\n{df.head()}")
```

### 4 Guardar el nuevo DF como CSV si es necesario

```python
# Guardar como nuevo archivo si es necesario
new_name = "limpieza_" + nombre_csv 
df.to_csv(new_name, index=False, sep=';')
```
