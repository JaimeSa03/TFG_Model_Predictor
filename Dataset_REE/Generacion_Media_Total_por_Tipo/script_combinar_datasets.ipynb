```python
import pandas as pd
import glob

# Buscar todos los archivos CSV relevantes
archivos_csv1 = [
    "DemandaReal_Enero_Abril_2020_peninsular.csv",
    "DemandaReal_Mayo_Agosto_2020_peninsular.csv", 
    "DemandaReal_Sept_Diciem_2020_peninsular.csv"
]

#busca los archivos csv que empiecen por Generacion
archivos_csv = glob.glob("limpieza*.csv") 

# Combinar archivos
dataframes = []
for archivo in archivos_csv:
    df = pd.read_csv(archivo, sep=';')
    dataframes.append(df)

# Crear DataFrame unico
df_completo = pd.concat(dataframes, ignore_index=True)

# Ordenar la fecha, si existe
if 'datetime' in df_completo.columns:
    df_completo['datetime'] = pd.to_datetime(df_completo['datetime'])
    df_completo = df_completo.sort_values('datetime')

# Guardarlo en un CSV
df_completo.to_csv("GeneracionMediaTipos2020_2024.csv", sep=';', index=False)
print(f"Archivos combinados:\n\n{archivos_csv1}")
print(f"Registros(filas) totales: {len(df_completo)}")
```
