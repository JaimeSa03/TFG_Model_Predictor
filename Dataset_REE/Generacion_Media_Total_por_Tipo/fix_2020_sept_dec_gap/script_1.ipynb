```python
import pandas as pd

# =========================
# 1. CARGAR DATASETS
# =========================

# Dataset original (long)
df_original = pd.read_csv("GeneracionMediaTipos2020_2024.csv", sep=";")

# Dataset Q4 (wide)
df_q4 = pd.read_csv("Limpiezageneracion_detallada_peninsula_2020Q4.csv")


# =========================
# 2. NORMALIZAR ORIGINAL
# =========================

df_original["datetime"] = pd.to_datetime(df_original["datetime"], utc=True)


# =========================
# 3. CONVERTIR Q4 A LONG
# =========================

df_q4_long = df_q4.melt(
    id_vars=["fecha"],
    var_name="name",
    value_name="value"
)

# Renombrar fecha → datetime
df_q4_long = df_q4_long.rename(columns={"fecha": "datetime"})

# Convertir a datetime (importante por zona horaria)
df_q4_long["datetime"] = pd.to_datetime(df_q4_long["datetime"], utc=True)


# =========================
# 4. MAPEO DE NOMBRES
# =========================

mapping = {
    "biomasa": "Generación medida Biomasa",
    "ciclo_combinado": "Generación medida Ciclo combinado",
    "hidraulica_ugh": "Generación medida Hidráulica UGH",
    "hidraulica_no_ugh": "Generación medida Hidráulica no UGH",
    "eolica_terrestre": "Generación medida Eólica terrestre",
    "nuclear": "Generación medida Nuclear",
    "solar_fotovoltaica": "Generación medida Solar fotovoltaica",
    "solar_termica": "Generación medida Solar térmica",
    "biogas": "Generación medida Biogás",
    "residuos_varios": "Generación medida Residuos varios",
    "residuos_domesticos_y_similares": "Generación medida Residuos domésticos y similares",
    "subproductos_mineria": "Generación medida Subproductos minería",
    "turbinacion_bombeo": "Generación medida Turbinación bombeo",
    "derivados_del_petroleo_o_carbon": "Generación medida Derivados del petróleo ó carbón",
    "energia_residual": "Generación medida Energía residual",
    "gas_natural_cogeneracion": "Generación medida Gas Natural Cogeneración",
    "oceano_y_geotermica": "Generación medida Océano y geotérmica",
    "hulla_antracita": "Generación medida Hulla antracita",
    "hulla_sub_bituminosa": "Generación medida Hulla sub-bituminosa",
    "fuel": "Generación medida Fuel"
}

# Aplicar mapping
#df_q4_long["name"] = df_q4_long["name"].map(mapping)


# =========================
# 5. LIMPIEZA Q4
# =========================

# Eliminar filas sin nombre (columnas que no estaban en mapping)
#df_q4_long = df_q4_long.dropna(subset=["name"])

# Eliminar valores nulos
#df_q4_long = df_q4_long.dropna(subset=["value"])


# =========================
# 6. UNIR DATASETS
# =========================

df_full = pd.concat([df_original, df_q4_long], ignore_index=True)


# =========================
# 7. ELIMINAR DUPLICADOS
# =========================

df_full = df_full.drop_duplicates(
    subset=["datetime", "name"],
    keep="last"
)


# =========================
# 8. ORDENAR
# =========================

df_full = df_full.sort_values("datetime")


# =========================
# 9. GUARDAR
# =========================

df_full.to_csv("GeneracionTipos2020_2024_COMPLETO.csv", index=False, sep=';')

print("✅ Dataset unificado guardado correctamente")
print(df_full.head())
```
