```python
import os
import pandas as pd

# 1. Configuración de rutas y años
# Reemplaza con la ruta absoluta de tu carpeta principal si es necesario
ruta_base = r"."
anos = range(2020, 2025)  # Genera [2020, 2021, 2022, 2023, 2024]

# Lista para almacenar los DataFrames de cada año
lista_dataframes = []

# 2. Bucle para leer y validar cada archivo
for ano in anos:
    ruta_carpeta = os.path.join(ruta_base, str(ano))
    
    # Comprobar si la carpeta del año existe
    if os.path.exists(ruta_carpeta):
        # Buscar archivos .csv.gz en la carpeta
        archivos = [f for f in os.listdir(ruta_carpeta) if f.endswith('.csv.gz')]
        
        for archivo in archivos:
            ruta_completa = os.path.join(ruta_carpeta, archivo)
            print(f"Procesando: {ruta_completa}")
            
            # Pandas detecta automáticamente la compresión .gz
            df = pd.read_csv(ruta_completa)
            
            # Convertir la columna 'time' a formato datetime para asegurar homogeneidad
            df['time'] = pd.to_datetime(df['time'])
            
            lista_dataframes.append(df)
    else:
        print(f"Advertencia: La carpeta para el año {ano} no existe en {ruta_base}")

# 3. Concatenar todos los años en un único Súper Dataset
if lista_dataframes:
    # Unión vertical (stacking) de todos los datos
    super_dataset = pd.concat(lista_dataframes, ignore_index=True)
    
    # Ordenar cronológicamente por la columna 'time'
    super_dataset = super_dataset.sort_values(by='time').reset_index(drop=True)
    
    # Formatear la columna 'time' al formato exacto requerido: 'datetime' AAAA-MM-DD HH:MM:SS
    super_dataset['time'] = super_dataset['time'].dt.strftime('%Y-%m-%d %H:%M:%S')
    super_dataset = super_dataset.rename(columns={'time': 'datetime'})
    
    # 4. Guardar el resultado finalS
    ruta_salida = os.path.join('./', "super_dataset_climatologia.csv")
    super_dataset.to_csv(ruta_salida, index=False)
    
    print(f"\n¡Éxito! Súper dataset creado con {super_dataset.shape[0]} filas y guardado en: {ruta_salida}")
else:
    print("\nNo se encontraron archivos para procesar.")
```
