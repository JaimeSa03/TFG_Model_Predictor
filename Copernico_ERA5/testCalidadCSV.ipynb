 2. Temporal # Que tienes las 8.784h de 2020 sin huecos
 3. Nulos # Que la limpieza funcionó bien
 4. Rangos físicos # Que la temperatura no es -500K ni el viento 900 m/s
 5. Geografía # Que todos los puntos están en España

 7. Estadísticas # Resumen numérico completo

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import os

from files.validacion import ES_BISIESTO

# ============================================================
# CONFIGURACIÓN GENERAL
# ============================================================
YEARS     = range(2020, 2025) 
# YEARS     = range(2025, 2026)
BASE_PATH = "."


# Umbrales térmicos - deben coincidir con pipeline.py
T_REF_FRIO  = 19.0
T_REF_CALOR = 28.0

# ============================================================
# COLUMNAS ESPERADAS EN EL CSV
# ============================================================
COLUMNAS_ESPERADAS = [
    'time',
    # [NUEVO] Demanda térmica - 46 capitales colapsadas en 3 columnas
    'hdd_peninsular_ponderado',     # estrés por frío (pesos INE dinámicos)
    'cdd_peninsular_ponderado',     # estrés por calor (pesos INE dinámicos)
    'estres_termico_total',         # HDD + CDD (U-shape resuelta)
    # Solar - ponderado por potencia instalada del año
    'ssrd_solar_ponderado',         # radiación solar (J/m²)
    'tcc_solar_ponderado',          # nubosidad (0-1)
    't2m_celsius_solar_ponderado',  # temperatura solar (°C)
    # Eólica - ponderado por potencia instalada del año
    'wind_speed_100m_ponderado',    # velocidad viento 100m (m/s)
    'wind_speed_10m_ponderado',     # velocidad viento 10m (m/s)
    'wind_shear_ponderado',         # gradiente vertical (ratio, capeado a 5)
    # Hidráulica - ponderado por potencia instalada (fija)
    'tp_hidro_ponderado',           # precipitación acumulada (m/h)
]

# ============================================================
# RANGOS FÍSICOS ESPERADOS PARA ESPAÑA
# Rangos para HDD/CDD horarios (escala /24)
# ============================================================
RANGOS = {
    # HDD/CDD - escala horaria (°día/h = °día / 24)
    'hdd_peninsular_ponderado': (
        0, T_REF_FRIO / 24, "°día/h",
        "HDD peninsular ponderado"
    ),
    'cdd_peninsular_ponderado': (
        0, 30 / 24, "°día/h",
        "CDD peninsular ponderado"
    ),
    'estres_termico_total': (
        0, (T_REF_FRIO + 30) / 24, "°día/h",
        "Estrés térmico total"
    ),
    # Solar
    'ssrd_solar_ponderado':        (0,   4e6,  "J/m²", "Radiación solar ponderada"),
    'tcc_solar_ponderado':         (0,   1,    "-",    "Nubosidad ponderada"),
    't2m_celsius_solar_ponderado': (-15, 48,   "°C",   "Temperatura solar ponderada"),
    # Eólica
    'wind_speed_100m_ponderado':   (0,   60,   "m/s",  "Viento 100m ponderado"),
    'wind_speed_10m_ponderado':    (0,   40,   "m/s",  "Viento 10m ponderado"),
    'wind_shear_ponderado':        (0,   5,    "-",    "Wind shear ponderado"),
    # Hidráulica
    'tp_hidro_ponderado':          (0,   0.15, "m",    "Precipitación hidráulica ponderada"),
}

def procesar_año(year):

    ES_BISIESTO     = (year % 4 == 0)
    HORAS_ESPERADAS = 366 * 24 if ES_BISIESTO else 365 * 24

    # ============================================================
    # CARGA
    # ============================================================
    carpeta = os.path.join(BASE_PATH, str(year))
    print(f"\n{'='*60}")
    print(f"Validación año {year}")
    print(f"{'='*60}")

    path = os.path.join(carpeta, f"dataset_{year}.csv.gz")
    if not os.path.exists(path):
        print(f"  ❌ No encontrado: {path}")
        return None
    
    df = pd.read_csv(path, compression='gzip')
    df['time'] = pd.to_datetime(df['time'])
    print(f"✅ Dataset cargado - Shape: {df.shape}")

    # ============================================================
    # BLOQUE 1 - ESTRUCTURA
    # ============================================================
    print(f"\n{'='*50}")
    print(f"📋 BLOQUE 1 - ESTRUCTURA")
    print(f"{'='*50}")

    print(f"\n🔢 Filas:    {df.shape[0]:,}")
    print(f"🔢 Columnas: {df.shape[1]}")
    print(f"\n📌 Columnas presentes:\n{df.columns.tolist()}")
    print(f"\n📌 Tipos de datos:\n{df.dtypes}")

    faltantes = [c for c in COLUMNAS_ESPERADAS if c not in df.columns]
    extras    = [c for c in df.columns if c not in COLUMNAS_ESPERADAS]

    if faltantes:
        print(f"\n⚠️  Columnas FALTANTES: {faltantes}")
    else:
        print(f"\n✅ Todas las columnas esperadas presentes")
    if extras:
        print(f"ℹ️ Columnas extra (no esperadas): {extras}")
        
    # ============================================================
    # BLOQUE 2 - COBERTURA TEMPORAL
    # ============================================================
    print(f"\n{'='*50}")
    print(f"🕐 BLOQUE 2 - COBERTURA TEMPORAL")
    print(f"{'='*50}")

    print(f"\n📅 Inicio:  {df['time'].min()}")
    print(f"📅 Fin:     {df['time'].max()}")
    print(f"📅 Horas:   {df['time'].nunique():,} / {HORAS_ESPERADAS:,} esperadas")

    if df['time'].nunique() == HORAS_ESPERADAS:
        print("✅ Cobertura temporal completa")
    else:
        diff = HORAS_ESPERADAS - df['time'].nunique()
        print(f"⚠️  Faltan {diff} horas")
        todas  = pd.date_range(f'{year}-01-01', f'{year}-12-31 23:00', freq='h')
        huecos = todas.difference(df['time'])
        if len(huecos):
            print(f"   Primeros huecos: {huecos[:5].tolist()}")

    # Verificar continuidad - no debe haber saltos > 1h
    df_s   = df.sort_values('time')
    diffs  = df_s['time'].diff().dropna()
    saltos = diffs[diffs != pd.Timedelta('1h')]
    if saltos.empty:
        print("✅ Serie temporal continua, sin saltos")
    else:
        print(f"⚠️  {len(saltos)} saltos detectados")
        print(f"   Primeros: {saltos.head(3)}")

    # ============================================================
    # BLOQUE 3 - NULOS Y DUPLICADOS
    # ============================================================
    print(f"\n{'='*50}")
    print(f"🔍 BLOQUE 3 - NULOS Y DUPLICADOS")
    print(f"{'='*50}")

    nulos     = df.isnull().sum()
    nulos_pct = (nulos / len(df) * 100).round(2)
    tabla_nulos = pd.DataFrame({'nulos': nulos, '%': nulos_pct})
    tabla_nulos = tabla_nulos[tabla_nulos['nulos'] > 0]

    if tabla_nulos.empty:
        print("\n✅ Sin valores nulos")
    else:
        print(f"\n⚠️  Columnas con nulos:\n{tabla_nulos}")

    dups = df.duplicated(subset=['time']).sum()
    print(f"\n{'✅' if dups == 0 else '⚠️'} Filas duplicadas por time: {dups}")

    # ============================================================
    # BLOQUE 4 - RANGOS FÍSICOS
    # ============================================================
    print(f"\n{'='*50}")
    print(f"🌡️  BLOQUE 4 - RANGOS FÍSICOS")
    print(f"{'='*50}")

    for col, (vmin, vmax, unidad, nombre) in RANGOS.items():
        if col not in df.columns:
            print(f"\n  ⚠️  {nombre} ({col}) - columna no encontrada")
            continue
        rmin = df[col].min()
        rmax = df[col].max()
        ok   = rmin >= vmin and rmax <= vmax
        print(f"\n  {'✅' if ok else '⚠️ '} {nombre}")
        print(f"     Esperado: [{vmin}, {vmax}] {unidad}")
        print(f"     Real:     [{rmin:.4f}, {rmax:.4f}] {unidad}")

    # ============================================================
    # BLOQUE 5 - VISUALIZACIONES
    # [NUEVO] Panel adaptado a las nuevas columnas HDD/CDD
    # ============================================================
    print(f"\n{'='*50}")
    print(f"📊 BLOQUE 5 - VISUALIZACIONES")
    print(f"{'='*50}")

    fig = plt.figure(figsize=(20, 20))
    fig.suptitle(f'Validación Dataset ERA5 - España {year}',
                fontsize=14, fontweight='bold')
    gs = gridspec.GridSpec(4, 3, figure=fig)

    # -- Fila 1: HDD y CDD diarios ------------------------------

    # [NUEVO] HDD peninsular ponderado - acumulado diario
    ax1 = fig.add_subplot(gs[0, :2])
    if 'hdd_peninsular_ponderado' in df.columns:
        hdd_d = df.groupby(df['time'].dt.date)['hdd_peninsular_ponderado'].sum()
        ax1.fill_between(hdd_d.index, hdd_d.values, color='steelblue', alpha=0.6)
        ax1.plot(hdd_d.index, hdd_d.values, color='steelblue', linewidth=0.6)
    ax1.set_title('HDD peninsular ponderado diario\n'
                f'(pesos INE {year} - estrés por frío, acumulado horario)')
    ax1.set_ylabel('°día acum.')
    ax1.grid(alpha=0.3)

    # [NUEVO] CDD mensual
    ax2 = fig.add_subplot(gs[0, 2])
    if 'cdd_peninsular_ponderado' in df.columns:
        cdd_m = df.groupby(df['time'].dt.month)['cdd_peninsular_ponderado'].sum()
        ax2.bar(cdd_m.index, cdd_m.values, color='tomato')
    ax2.set_title('CDD mensual\n(estrés por calor)')
    ax2.set_xlabel('Mes')
    ax2.grid(alpha=0.3, axis='y')

    # -- Fila 2: Estrés térmico total y scatter HDD vs CDD ------

    # [NUEVO] Estrés térmico total diario - U-shape resuelta
    ax3 = fig.add_subplot(gs[1, :2])
    if 'estres_termico_total' in df.columns:
        et_d = df.groupby(df['time'].dt.date)['estres_termico_total'].sum()
        ax3.plot(et_d.index, et_d.values, color='purple', linewidth=0.7)
        ax3.fill_between(et_d.index, et_d.values, color='purple', alpha=0.15)
    ax3.set_title('Estrés térmico total diario (HDD + CDD)\n'
                'El modelo ve la U-shape completa en una sola variable')
    ax3.set_ylabel('°día acum.')
    ax3.grid(alpha=0.3)

    # [TODO] HDD vs CDD scatter
    # ¿Si un punto tiene HDD alto, CDD debe ser 0 y viceversa?
    ax4 = fig.add_subplot(gs[1, 2])
    if 'hdd_peninsular_ponderado' in df.columns and \
    'cdd_peninsular_ponderado' in df.columns:
        # Muestrear para no sobrecargar el scatter
        df_sample = df.sample(min(2000, len(df)), random_state=42)
        ax4.scatter(df_sample['hdd_peninsular_ponderado'],
                    df_sample['cdd_peninsular_ponderado'],
                    alpha=0.2, s=2, color='purple')
    ax4.set_xlabel('HDD (frío)')
    ax4.set_ylabel('CDD (calor)')
    ax4.set_title('HDD vs CDD\n')
    ax4.grid(alpha=0.3)

    # -- Fila 3: Solar y viento ----------------------------------

    # Radiación solar ponderada diaria
    ax5 = fig.add_subplot(gs[2, :2])
    if 'ssrd_solar_ponderado' in df.columns:
        rad_d = df.groupby(df['time'].dt.date)['ssrd_solar_ponderado'].mean() / 1e6
        ax5.plot(rad_d.index, rad_d.values, color='goldenrod', linewidth=0.7)
        ax5.fill_between(rad_d.index, rad_d.values, color='goldenrod', alpha=0.2)
    ax5.set_title('Radiación solar media diaria ponderada (MJ/m²)\n'
                'Esperado: campana con máximo en junio/julio')
    ax5.set_ylabel('MJ/m²')
    ax5.grid(alpha=0.3)

    # Viento 100m ponderado diario
    ax6 = fig.add_subplot(gs[2, 2])
    if 'wind_speed_100m_ponderado' in df.columns:
        v_d = df.groupby(df['time'].dt.date)['wind_speed_100m_ponderado'].mean()
        ax6.plot(v_d.index, v_d.values, color='seagreen', linewidth=0.7)
    ax6.set_title('Viento 100m\nponderado (m/s)')
    ax6.grid(alpha=0.3)

    # -- Fila 4: Precipitación + correlación ---------------------

    # Precipitación hidráulica mensual
    ax7 = fig.add_subplot(gs[3, 0])
    if 'tp_hidro_ponderado' in df.columns:
        tp_m = df.groupby(df['time'].dt.month)['tp_hidro_ponderado'].sum() * 1000
        ax7.bar(tp_m.index, tp_m.values, color='steelblue')
    ax7.set_title('Precipitación mensual\nhidráulica ponderada (mm)')
    ax7.set_xlabel('Mes')
    ax7.grid(alpha=0.3, axis='y')

    # Temperatura solar ponderada diaria
    ax8 = fig.add_subplot(gs[3, 1])
    if 't2m_celsius_solar_ponderado' in df.columns:
        t_d = df.groupby(df['time'].dt.date)['t2m_celsius_solar_ponderado'].mean()
        ax8.plot(t_d.index, t_d.values, color='tomato', linewidth=0.7)
        ax8.axhline(y=T_REF_FRIO,  color='steelblue', linestyle='--',
                    alpha=0.5, label=f'{T_REF_FRIO}°C HDD ref')
        ax8.axhline(y=T_REF_CALOR, color='tomato',    linestyle='--',
                    alpha=0.5, label=f'{T_REF_CALOR}°C CDD ref')
        ax8.legend(fontsize=7)
    ax8.set_title('Temperatura solar ponderada\ndiaria (°C)')
    ax8.set_ylabel('°C')
    ax8.grid(alpha=0.3)

    # [NUEVO] Matriz de correlación - variables clave del modelo
    ax9 = fig.add_subplot(gs[3, 2])
    cols_corr = [c for c in [
        'hdd_peninsular_ponderado',
        'cdd_peninsular_ponderado',
        'estres_termico_total',
        'ssrd_solar_ponderado',
        'tcc_solar_ponderado',
        'wind_speed_100m_ponderado',
        'wind_shear_ponderado',
        'tp_hidro_ponderado',
    ] if c in df.columns]

    if cols_corr:
        corr = df[cols_corr].corr()
        im   = ax9.imshow(corr, cmap='RdBu_r', vmin=-1, vmax=1)
        lbls = [c.replace('_peninsular_ponderado', '')
                .replace('_ponderado', '')
                .replace('wind_speed_', 'v')
                .replace('t2m_celsius_', 't_')
                for c in cols_corr]
        ax9.set_xticks(range(len(cols_corr)))
        ax9.set_yticks(range(len(cols_corr)))
        ax9.set_xticklabels(lbls, rotation=45, ha='right', fontsize=6)
        ax9.set_yticklabels(lbls, fontsize=6)
        ax9.set_title('Correlación\nvariables clave')
        plt.colorbar(im, ax=ax9)

    plt.tight_layout()
    output_img = f"{year}/validacion_{year}.png"
    plt.savefig(output_img, dpi=150, bbox_inches='tight')
    plt.show()
    print(f"  ✅ Gráfico guardado: {output_img}")

    # ============================================================
    # BLOQUE 6 - RESUMEN ESTADÍSTICO
    # ============================================================
    print(f"\n{'='*50}")
    print(f"📊 BLOQUE 6 - RESUMEN ESTADÍSTICO")
    print(f"{'='*50}")

    # Resumen térmico
    print(f"\n🌡️  TÉRMICO (pesos INE {year}):")
    cols_t = ['hdd_peninsular_ponderado',
            'cdd_peninsular_ponderado',
            'estres_termico_total']
    cols_t = [c for c in cols_t if c in df.columns]
    if cols_t:
        print(df[cols_t].describe().round(6).to_string())

        # -- Análisis de solapamiento HDD/CDD ---------------------
        # El solapamiento NO es un error - es física real de España:
        # mientras Soria tiene calefacción, Sevilla tiene A/C.
        # Ocurre en horas de transición (primavera, otoño, madrugadas)
        # donde el gradiente térmico norte/sur es máximo.

        if all(c in df.columns for c in
            ['hdd_peninsular_ponderado', 'cdd_peninsular_ponderado']):

            mascara_solapamiento = (
                (df['hdd_peninsular_ponderado'] > 0) &
                (df['cdd_peninsular_ponderado'] > 0)
            )
            horas_solapamiento = mascara_solapamiento.sum()

            print(f"\n  ℹ️  Horas con HDD>0 y CDD>0 simultáneamente: "
                f"{horas_solapamiento:,} "
                f"({horas_solapamiento/len(df)*100:.1f}% del año)")
            print(f"  ✅ Esto es física real - norte frío + sur cálido a la vez")
            print(f"     El modelo aprende este gradiente térmico peninsular")

            # Distribución mensual del solapamiento
            # ¿En qué meses coexisten calefacción (norte) y A/C (sur)?
            df_solap = df[mascara_solapamiento].copy()
            if not df_solap.empty:
                por_mes = df_solap.groupby(
                    df_solap['time'].dt.month
                ).size()

                nombres_mes = {
                    1:'Ene', 2:'Feb', 3:'Mar', 4:'Abr',
                    5:'May', 6:'Jun', 7:'Jul', 8:'Ago',
                    9:'Sep', 10:'Oct', 11:'Nov', 12:'Dic'
                }

                print(f"\n  📅 Distribución mensual del solapamiento "
                    f"(norte frío + sur cálido simultáneo):")
                print(f"  {'Mes':<6} {'Horas':>6} {'% del mes':>10} "
                    f"{'Barra visual'}")
                print(f"  {'-'*55}")

                for mes, horas in por_mes.items():
                    # Horas totales del mes para calcular porcentaje
                    horas_mes = (
                        df['time'].dt.month == mes
                    ).sum()
                    pct = horas / horas_mes * 100
                    barra = '█' * int(pct / 2)
                    print(f"  {nombres_mes[mes]:<6} {horas:>6,} "
                        f"{pct:>9.1f}%  {barra}")

                # Meses con más solapamiento - corroboran la hipótesis
                mes_max = por_mes.idxmax()
                mes_min = por_mes[por_mes > 0].idxmin() \
                        if (por_mes > 0).any() else None

                print(f"\n  🔝 Mes con más solapamiento:   "
                    f"{nombres_mes[mes_max]} "
                    f"({por_mes[mes_max]:,}h) "
                    f"← transición térmica máxima")
                if mes_min:
                    print(f"  🔽 Mes con menos solapamiento: "
                        f"{nombres_mes[mes_min]} "
                        f"({por_mes[mes_min]:,}h) "
                        f"← clima más homogéneo")

                # Verificar hipótesis: ¿el solapamiento es mayor en
                # meses de transición (mar,abr,may,sep,oct,nov)?
                meses_transicion = [3, 4, 5, 9, 10, 11]
                meses_extremo    = [1, 2, 6, 7, 8, 12]

                h_transicion = por_mes[
                    por_mes.index.isin(meses_transicion)
                ].sum()
                h_extremo = por_mes[
                    por_mes.index.isin(meses_extremo)
                ].sum()

                print(f"\n  📊 Verificación hipótesis climática:")
                print(f"     Meses de transición "
                    f"(Mar-May, Sep-Nov): {h_transicion:,}h "
                    f"({h_transicion/horas_solapamiento*100:.1f}%)")
                print(f"     Meses extremos "
                    f"(Ene-Feb, Jun-Ago, Dic): {h_extremo:,}h "
                    f"({h_extremo/horas_solapamiento*100:.1f}%)")

                if h_transicion > h_extremo:
                    print(f"     ✅ Confirmado: el solapamiento es mayor "
                        f"en meses de transición")
                else:
                    print(f"     ℹ️  El solapamiento se distribuye de "
                        f"forma atípica - año {year} inusual")

                # Análisis por hora del día - ¿a qué hora ocurre más?
                por_hora = df_solap.groupby(
                    df_solap['time'].dt.hour
                ).size()

                hora_pico = por_hora.idxmax()
                print(f"\n  🕐 Hora del día con más solapamiento: "
                    f"{hora_pico:02d}:00h "
                    f"({por_hora[hora_pico]:,} ocurrencias)")
                print(f"     {'Madrugada (0-6h)' if hora_pico < 6 else 'Mañana (6-12h)' if hora_pico < 12 else 'Tarde (12-18h)' if hora_pico < 18 else 'Noche (18-24h)'}"
                    f" - cuando el gradiente N/S es más pronunciado")

    # Resumen energético
    print("\n⚡ ENERGÉTICO:")
    cols_e = ['ssrd_solar_ponderado', 'tcc_solar_ponderado',
            't2m_celsius_solar_ponderado',
            'wind_speed_100m_ponderado', 'wind_speed_10m_ponderado',
            'wind_shear_ponderado', 'tp_hidro_ponderado']
    cols_e = [c for c in cols_e if c in df.columns]
    if cols_e:
        print(df[cols_e].describe().round(4).to_string())

    print(f"\n{'='*60}")
    print(f"\n✅ Validación {year} completada")
    print("="*60)
    
# ============================================================
# EJECUCIÓN
# ============================================================
if __name__ == "__main__":

    for year in YEARS:
        shape = procesar_año(year)
```
