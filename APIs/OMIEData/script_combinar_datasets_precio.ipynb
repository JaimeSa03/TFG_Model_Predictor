# Resumen del script

Este script busca todos los archivos CSV que comienzan con Omie y termina en .csv, los lee en un bucle, los combina en un único DataFrame y los ordena por la columna DATE si existe.

Luego guarda el resultado en un archivo llamado OmiePrecio2020_2024.csv y muestra en consola:

- los archivos que se procesaron
- el número total de registros finales

Se ha empleado para los archivos del precio correspondientes para cada año, por ejemplo: Omie_precio2024.csv.

```python
import pandas as pd
import glob

# =========================
# Unir los archivos csv que empiecen por Omie y terminen en .csv
# =========================
archivos_csv = glob.glob("Omie*.csv") 

# Combinar archivos
dataframes = []
for archivo in archivos_csv:
    df = pd.read_csv(archivo, sep=';')
    dataframes.append(df)

# Crear DataFrame unico
df_completo = pd.concat(dataframes, ignore_index=True)

# Ordenar la fecha, si existe
if 'DATE' in df_completo.columns:
    df_completo['DATE'] = pd.to_datetime(df_completo['DATE'])
    df_completo = df_completo.sort_values('DATE')

# =========================
# Guardarlo en un CSV
# =========================

df_completo.to_csv("OmiePrecio2020_2024.csv", sep=';', index=False)
print(f"Archivos combinados:\n\n{archivos_csv}")
print(f"Registros(filas) totales: {len(df_completo)}")
```
