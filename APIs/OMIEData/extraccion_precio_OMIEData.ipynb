# Código para la descarga de datos precio de la luz

Descarga de los precios diarios de la electricidad correspondientes al mercado operado por OMIE mediante el uso de la API de OMIEData.

Devuelve el conjunto de datos dado por la fecha elegida. Por ejemplo: Omie_precio2024.csv.

Se descargaron los archivos de cada año de forma secuencial para evitar sobrecarga.

```python
import datetime as dt
import matplotlib.pyplot as plt

from OMIEData.DataImport.omie_marginalprice_importer import OMIEMarginalPriceFileImporter
from OMIEData.Enums.all_enums import DataTypeInMarginalPriceFile

# Seleccionamos la fecha
dateIni = dt.datetime(2024, 1, 1)
dateEnd = dt.datetime(2024, 12, 31)

# This can take time, it is downloading the files from the website..
df = OMIEMarginalPriceFileImporter(date_ini=dateIni, date_end=dateEnd).read_to_dataframe(verbose=True)
df.sort_values(by='DATE', axis=0, inplace=True)
print(df.head(8))
```

### Filtramos para obtener únicamente los datos pertenecientes al conjunto español

```python

# Precio solo de españa
str_price_spain = str(DataTypeInMarginalPriceFile.PRICE_SPAIN)
dfPrices = df[df.CONCEPT == str_price_spain]

print(dfPrices.head(10))  # Primeras 10 filas

# Plotting
plt.figure()
plt.plot(dfPrices.DATE, dfPrices.H8, label='Precio luz 8:00')
plt.plot(dfPrices.DATE, dfPrices.H20, label='Precio luz 20:00 pm')
plt.xticks(rotation=45)  # Gira las fechas 45 grados para poder leerlas bien
plt.legend()
plt.show()

# Guardamos el df en un archivo csv
dfPrices.to_csv("Omie_precio2024.csv", index=False, sep=';')
```
