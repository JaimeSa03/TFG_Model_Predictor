### 1. ¿Qué es una serie temporal y por qué es especial?
Una serie temporal es una secuencia de datos ordenados en el tiempo, por ejemplo:

- Precio de la luz cada hora

- Precio medio diario

- Precio medio mensual

Hay por lo tanto un aspecto clave y es precisamente la evolución de estos datos con el tiempo, no tratamos con sucesos aleatorios de los cuales no podríamos extraer mucha información. Con una serie temporal se trata de analizar lo que ha ocurrido en el pasado para poder predecir lo que ocurrirá el futuro.



### 2. Relación entre el precio de la luz y las Series Temporales
El precio de la electricidad en España es una serie temporal clara porque:

- El orden temporal importa (no puedes mezclar horas/días).

- Hay patrones repetitivos (horarios, semanales, estacionales).

- El valor futuro depende de los valores pasados.

Esto hace que modelos clásicos de regresión no sean tan adecuados, aquí entran en juego modelos como ARIMA y SARIMA.

### 3. Resumen de la idea intuitiva de estos modelos
ARIMA y SARIMA se basan en una idea muy sencilla:

"El precio de hoy se puede explicar como una combinación del precio de ayer, de los errores pasados y de la tendencia general"

No intentan entender el mercado eléctrico como tal, sino explotar la estructura temporal del dato.
